```python
import pandas as pd

df = pd.read_csv("pizzas.csv")
df
```

```python
df.plot(
    kind="scatter",
    x="diametro",
    y="preco",
    title="🍕 Relação entre diâmetro e preço"
);
```

```python
from sklearn import linear_model

modelo = linear_model.LinearRegression()
x = df[["diametro"]]
y = df[["preco"]]

modelo.fit(x, y)
```

```python
novo = pd.DataFrame({
    "diametro": [67]
})

modelo.predict(novo)
```

```python
import matplotlib.pyplot as plt

plt.scatter(
    df["diametro"],
    df["preco"]
)

plt.plot(
    df["diametro"],
    modelo.predict(x)
)

plt.xlabel("Diâmetro")
plt.ylabel("Preço")
plt.title("🍕 IA aprendendo preços")

plt.show()
```
