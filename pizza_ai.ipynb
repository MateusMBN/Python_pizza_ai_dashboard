```python
import pandas as pd

df = pd.read_csv("pizzas.csv")

df
```

```python
from sklearn.linear_model import LinearRegression

# Entradas
x = df[[
    "diametro",
    "ingredientes",
    "borda"
]]

# Saída
y = df["preco"]

# Modelo
modelo = LinearRegression()

# Treinar
modelo.fit(x, y)
```

```python
nova_pizza = pd.DataFrame({
    "diametro": [35],
    "ingredientes": [7],
    "borda": [1]
})

modelo.predict(nova_pizza)
```

```python
modelo.coef_
```

```python
import matplotlib.pyplot as plt

plt.scatter(
    df["diametro"],
    df["preco"]
)

plt.xlabel("Diâmetro")
plt.ylabel("Preço")
plt.title("🍕 Relação entre tamanho e preço")

plt.show()
```

```python
coeficientes = pd.DataFrame({
    "Variavel": x.columns,
    "Peso": modelo.coef_
})

coeficientes
```

```python
previsoes = modelo.predict(x)

resultado = df.copy()

resultado["previsao_ia"] = previsoes

resultado
```

```python
modelo.score(x, y)
```

```python
from sklearn.tree import DecisionTreeRegressor
```

```python
modelo_arvore = DecisionTreeRegressor()

modelo_arvore.fit(x, y)
```

```python
nova_pizza = pd.DataFrame({
    "diametro": [35],
    "ingredientes": [7],
    "borda": [1]
})

modelo_arvore.predict(nova_pizza)
```

```python
print("Linear Regression:")
print(modelo.predict(nova_pizza))

print()

print("Decision Tree:")
print(modelo_arvore.predict(nova_pizza))
```

```python
from sklearn import tree
import matplotlib.pyplot as plt

plt.figure(figsize=(15,10))

tree.plot_tree(
    modelo_arvore,
    feature_names=x.columns,
    filled=True
)

plt.show()
```

```python
from sklearn.ensemble import RandomForestRegressor
```

```python
modelo_floresta = RandomForestRegressor(
    n_estimators=100,
    random_state=42
)

modelo_floresta.fit(x, y)
```

```python
nova_pizza = pd.DataFrame({
    "diametro": [35],
    "ingredientes": [7],
    "borda": [1]
})

modelo_floresta.predict(nova_pizza)
```

```python
print("📈 Linear Regression")
print(modelo.predict(nova_pizza))

print()

print("🌲 Decision Tree")
print(modelo_arvore.predict(nova_pizza))

print()

print("🌳 Random Forest")
print(modelo_floresta.predict(nova_pizza))
```

```python
print("Linear Regression:")
print(modelo.score(x, y))

print()

print("Decision Tree:")
print(modelo_arvore.score(x, y))

print()

print("Random Forest:")
print(modelo_floresta.score(x, y))
```

```python
from sklearn.ensemble import GradientBoostingRegressor
```

```python
modelo_boost = GradientBoostingRegressor(
    n_estimators=100,
    learning_rate=0.1,
    random_state=42
)

modelo_boost.fit(x, y)
```

```python
nova_pizza = pd.DataFrame({
    "diametro": [35],
    "ingredientes": [7],
    "borda": [1]
})

modelo_boost.predict(nova_pizza)
```

```python
print("📈 Linear Regression")
print(modelo.score(x, y))

print()

print("🌲 Decision Tree")
print(modelo_arvore.score(x, y))

print()

print("🌳 Random Forest")
print(modelo_floresta.score(x, y))

print()

print("🔥 Gradient Boosting")
print(modelo_boost.score(x, y))
```

```python
from xgboost import XGBRegressor
```
